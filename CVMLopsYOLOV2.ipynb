```python
#same as the cell for testing on cpu but with mlflow
import mlflow, os
from ultralytics import YOLO

mlflow.set_tracking_uri("file:./mlruns")
mlflow.set_experiment("car_detection_yolo")

with mlflow.start_run(run_name="yolov8n_colab"):

    model = YOLO("yolov8n.pt")
    results = model.train(
        data="data/yolo_dataset/car_detection.yaml",
        imgsz=320,
        epochs=1,
        batch=4,
        device="cpu",
        project="runs/train",
        name="car_yolo_v1",
        exist_ok=True
    )

    metrics = results.results_dict
    for k, v in metrics.items():
        try:
            mlflow.log_metric(k, float(v))
        except:
            pass

    mlflow.log_param("epochs", 20)
    mlflow.log_param("batch", 16)
    mlflow.log_param("model", "yolov8n")

    # log best weights as artifact
    mlflow.log_artifact("runs/train/car_yolo_v1/weights/best.pt")
```

```python
#GPU MLFLOW and Training with yolo v8 together (change runtime type to GPU)
import mlflow, os
from ultralytics import YOLO

mlflow.set_tracking_uri("file:./mlruns")
mlflow.set_experiment("car_detection_yolo")

with mlflow.start_run(run_name="yolov8n_colab"):

    model = YOLO("yolov8n.pt")
    results = model.train(
        data="data/yolo_dataset/car_detection.yaml",
        imgsz=640,
        epochs=20,
        batch=16,
        device=0,
        project="runs/train",
        name="car_yolo_v1",
        exist_ok=True
    )

    metrics = results.results_dict
    for k, v in metrics.items():
        try:
            mlflow.log_metric(k, float(v))
        except:
            pass

    mlflow.log_param("epochs", 20)
    mlflow.log_param("batch", 16)
    mlflow.log_param("model", "yolov8n")

    # log best weights as artifact
    mlflow.log_artifact("runs/train/car_yolo_v1/weights/best.pt")
```
